```python
#PREDICTIVE QUESTION: Can you predict the rating of a recipe
#based on its ingredients?
```

```python
#Loading data
import pandas as pd
from google.colab import files

uploaded = files.upload()
filename = next(iter(uploaded))
df = pd.read_csv(filename)

print("Rows and columns:", df.shape)
display(df.head())
```

```python
# Checks column names, data types, and missing values.
print("Column names:")
print(df.columns.tolist())

summary = pd.DataFrame({
    "data_type": df.dtypes,
    "missing_values": df.isna().sum(),
    "unique_values": df.nunique()
})
display(summary)
```

```python
# Keep recipes with both ingredient text and an observed rating.
model_df = df.dropna(subset=["ingredients", "avg_rating"]).copy()

# X is the information used to predict.
# y is the value we want to predict.
X = model_df["ingredients"]
y = model_df["avg_rating"]

print("Recipes available for modeling:", len(model_df))
print("\nExample ingredients:")
print(X.iloc[0])

print("\nRating summary:")
print(y.describe())
```
